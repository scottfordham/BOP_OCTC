```python
import os
from collections import Counter

import rasterio
import numpy as np
import matplotlib.pyplot as plt

# =============================================================================
# PATHS
# =============================================================================

base_folder = r"C:\NCA_DATA\Ancillary_Data\LTDL"

tile_paths = [
    r"C:\NCA_DATA\templates\grid_template_10m_26911_uint8.tif",
    r"C:\NCA_DATA\S2_Harmonics_vrt\S2SR_Harmonics_2025_4th_amp_phase_masked.tif",
    r"C:\NCA_DATA\Ancillary_Data\POLARIS_26911_10m_mosaic.tif",
    r"C:\NCA_DATA\Ancillary_Data\USGS3DEP_topo_26911_10m.tif",
    os.path.join(base_folder, "mgmt_binary_10m_new.tif"),
    os.path.join(base_folder, "mgmt_time_since_10m_new.tif"),
    os.path.join(base_folder, "mgmt_seeded_10m_new.tif"),
    os.path.join(base_folder, "mgmt_num_units_10m_new.tif"),
    os.path.join(base_folder, "mgmt_trt_major_10m_new.tif"),
    os.path.join(base_folder, "mgmt_trt_sub_10m_new.tif"),
    os.path.join(base_folder, "mgmt_treatment_type_10m_new.tif"),
    os.path.join(base_folder, "mgmt_most_recent_year_10m_new.tif"),
    os.path.join(base_folder, "mgmt_most_recent_time_since_10m_new.tif"),
    os.path.join(base_folder, "mgmt_times_treated_10m_new.tif"),
]

mgmt_rasters = [
    os.path.join(base_folder, "mgmt_binary_10m_new.tif"),
    os.path.join(base_folder, "mgmt_time_since_10m_new.tif"),
    os.path.join(base_folder, "mgmt_seeded_10m_new.tif"),
    os.path.join(base_folder, "mgmt_num_units_10m_new.tif"),
    os.path.join(base_folder, "mgmt_trt_major_10m_new.tif"),
    os.path.join(base_folder, "mgmt_trt_sub_10m_new.tif"),
    os.path.join(base_folder, "mgmt_treatment_type_10m_new.tif"),
    os.path.join(base_folder, "mgmt_most_recent_year_10m_new.tif"),
    os.path.join(base_folder, "mgmt_most_recent_time_since_10m_new.tif"),
    os.path.join(base_folder, "mgmt_times_treated_10m_new.tif"),
]

# =============================================================================
# HELPERS
# =============================================================================

def grid_signature(path):
    with rasterio.open(path) as ds:
        tr = ds.transform
        return (str(ds.crs), tr.a, tr.e, tr.c, tr.f, ds.width, ds.height)

def raster_summary(path):
    with rasterio.open(path) as ds:
        arr = ds.read(1, masked=True)
        valid = arr.compressed()

        summary = {
            "path": path,
            "shape": (ds.height, ds.width),
            "crs": str(ds.crs),
            "transform": ds.transform,
            "dtype": ds.dtypes[0],
            "nodata": ds.nodata,
            "valid_pixels": valid.size,
        }

        if valid.size > 0:
            summary.update({
                "min": float(valid.min()),
                "max": float(valid.max()),
                "mean": float(valid.mean()),
                "unique_preview": np.unique(valid)[:20]
            })
        else:
            summary.update({
                "min": None,
                "max": None,
                "mean": None,
                "unique_preview": []
            })

        return summary

def plot_raster(path, max_dim=2000):
    with rasterio.open(path) as ds:
        arr = ds.read(1, masked=True)

        # light downsampling for display only
        row_step = max(1, ds.height // max_dim)
        col_step = max(1, ds.width // max_dim)
        arr_small = arr[::row_step, ::col_step]

        plt.figure(figsize=(8, 6))
        plt.imshow(arr_small, interpolation="nearest")
        plt.title(os.path.basename(path))
        plt.colorbar()
        plt.tight_layout()
        plt.show()

# =============================================================================
# EXISTENCE CHECK
# =============================================================================

print("Checking raster existence...\n")
existing_tile_paths = []
for p in tile_paths:
    exists = os.path.exists(p)
    print(f"{exists}  {p}")
    if exists:
        existing_tile_paths.append(p)

print("\nChecking management rasters for visualization...\n")
existing_mgmt = []
for p in mgmt_rasters:
    exists = os.path.exists(p)
    print(f"{exists}  {p}")
    if exists:
        existing_mgmt.append(p)

# =============================================================================
# VISUALIZE + SUMMARIZE MANAGEMENT RASTERS
# =============================================================================

for p in existing_mgmt:
    print("\n" + "=" * 80)
    print(os.path.basename(p))
    info = raster_summary(p)
    for k, v in info.items():
        if k != "path":
            print(f"{k}: {v}")
    plot_raster(p)

# =============================================================================
# GRID SIGNATURE CHECK
# =============================================================================

print("\n" + "=" * 80)
print("GRID SIGNATURE CHECK")
print("=" * 80)

sigs = [grid_signature(p) for p in existing_tile_paths]
counts = Counter(sigs)

print("Unique grid signatures:", len(counts))
for sig, n in counts.most_common():
    print(n, sig)

# compare each file to template
template_path = r"C:\NCA_DATA\templates\grid_template_10m_26911_uint8.tif"
template_sig = grid_signature(template_path)

print("\n" + "=" * 80)
print("TEMPLATE COMPARISON")
print("=" * 80)

for p in existing_tile_paths:
    sig = grid_signature(p)
    print(f"{os.path.basename(p)}")
    print("Matches template:", sig == template_sig)
    if sig != template_sig:
        print("  Signature:", sig)
        print("  Template: ", template_sig)
    print()
```

```python
import os
import json
import shutil
import sys
from pathlib import Path

import numpy as np
import pandas as pd
import xarray as xr
import rasterio
import rioxarray as rxr

# =============================================================================
# SETTINGS
# =============================================================================

src_dir = Path(r"C:\NCA_DATA\Ancillary_Data\LTDL")
out_dir = Path(r"C:\NCA_DATA\Ancillary_Data\LTDL\LTDL_written")
out_dir.mkdir(parents=True, exist_ok=True)

# Existing rasters written from the LTDL workflow
rasters_to_copy = {
    "mgmt_binary": src_dir / "mgmt_binary_10m_new.tif",
    "mgmt_time_since": src_dir / "mgmt_time_since_10m_new.tif",
    "mgmt_seeded": src_dir / "mgmt_seeded_10m_new.tif",
    "mgmt_num_units": src_dir / "mgmt_num_units_10m_new.tif",
    "mgmt_trt_major": src_dir / "mgmt_trt_major_10m_new.tif",
    "mgmt_trt_sub": src_dir / "mgmt_trt_sub_10m_new.tif",
    "mgmt_treatment_type": src_dir / "mgmt_treatment_type_10m_new.tif",
    "mgmt_most_recent_year": src_dir / "mgmt_most_recent_year_10m_new.tif",
    "mgmt_most_recent_time_since": src_dir / "mgmt_most_recent_time_since_10m_new.tif",
    "mgmt_times_treated": src_dir / "mgmt_times_treated_10m_new.tif",
}

# If these CSVs already exist they will override the placeholder maps below.
csv_candidates = {
    "mgmt_trt_major": [
        src_dir / "lut_Trt_Type_Major_new.csv",
        out_dir / "lut_Trt_Type_Major_new.csv",
    ],
    "mgmt_trt_sub": [
        src_dir / "lut_Trt_Type_Sub_new.csv",
        out_dir / "lut_Trt_Type_Sub_new.csv",
    ],
    "mgmt_treatment_type": [
        src_dir / "lut_Treatment_Type_new.csv",
        out_dir / "lut_Treatment_Type_new.csv",
    ],
}

# Placeholder class maps. Replace these if you do not have CSV lookup files.
CLASS_MAPS = {
    "mgmt_binary": {
        1: "treated"
    },
    "mgmt_seeded": {
        0: "not_seeded",
        1: "seeded"
    },
    "mgmt_trt_major": {
        # 1: "Prescribed Burn",
        # 2: "Seeding",
        # 3: "Herbicide",
    },
    "mgmt_trt_sub": {
        # 1: "Aerial Seeding",
        # 2: "Drill Seeding",
    },
    "mgmt_treatment_type": {
        # 1: "Rangeland Drill Seeding",
        # 2: "Prescribed Burn",
    },
}

# Output filenames
nc_path = out_dir / "LTDL_management_stack.nc"
write_individual_netcdfs = False

# Set to True if you want to overwrite files in LTDL_written
overwrite_existing = True

# =============================================================================
# HELPERS
# =============================================================================

def log(msg):
    print(msg)

def ensure_absent(path: Path):
    if path.exists() and not overwrite_existing:
        raise RuntimeError(f"Output already exists and overwrite_existing=False:\n{path}")

def try_load_lookup_csv(csv_path: Path) -> dict[int, str]:
    df = pd.read_csv(csv_path)
    if df.shape[1] < 2:
        raise ValueError(f"Lookup CSV must have at least 2 columns:\n{csv_path}")
    code_col = df.columns[0]
    label_col = df.columns[1]
    out = {}
    for code, label in zip(df[code_col], df[label_col]):
        if pd.isna(code) or pd.isna(label):
            continue
        out[int(code)] = str(label)
    return out

def write_lookup_files(var_name: str, class_map: dict[int, str], folder: Path):
    if not class_map:
        return

    csv_path = folder / f"{var_name}_labels.csv"
    json_path = folder / f"{var_name}_labels.json"

    if csv_path.exists() and not overwrite_existing:
        raise RuntimeError(f"Lookup CSV already exists:\n{csv_path}")
    if json_path.exists() and not overwrite_existing:
        raise RuntimeError(f"Lookup JSON already exists:\n{json_path}")

    df = pd.DataFrame({
        "code": list(class_map.keys()),
        "label": list(class_map.values())
    }).sort_values("code")

    df.to_csv(csv_path, index=False)

    with open(json_path, "w", encoding="utf-8") as f:
        json.dump(class_map, f, indent=2)

    log(f"Wrote lookup CSV:  {csv_path}")
    log(f"Wrote lookup JSON: {json_path}")

def embed_classmap_in_tif(tif_path: Path, class_map: dict[int, str]):
    if not class_map:
        return
    with rasterio.open(tif_path, "r+") as dst:
        tags = dst.tags()
        tags["class_map_json"] = json.dumps(class_map)
        dst.update_tags(**tags)
    log(f"Embedded class_map_json into: {tif_path}")

def choose_dtype(var_name: str) -> str:
    """
    Keep all NetCDF variables float32 for simplicity and NaN handling.
    If you later want integer categories, we can tighten this.
    """
    return "float32"

# =============================================================================
# VERIFY netCDF4 BACKEND
# =============================================================================

log(f"Python executable: {sys.executable}")

try:
    import netCDF4
    log(f"netCDF4 version: {netCDF4.__version__}")
except ImportError as e:
    raise RuntimeError(
        "netCDF4 is not available in this Python environment.\n"
        "Use the environment where NLDAS writing worked or install netCDF4."
    ) from e

# =============================================================================
# LOAD LOOKUPS FROM CSV IF AVAILABLE
# =============================================================================

for key, candidates in csv_candidates.items():
    for csv_path in candidates:
        if csv_path.exists():
            CLASS_MAPS[key] = try_load_lookup_csv(csv_path)
            log(f"Loaded lookup for {key} from {csv_path}")
            break

# =============================================================================
# COPY TIFFS INTO LTDL_written
# =============================================================================

written_rasters = {}

for var_name, src_path in rasters_to_copy.items():
    if not src_path.exists():
        log(f"Missing source raster: {src_path}")
        continue

    dst_path = out_dir / src_path.name

    if dst_path.exists():
        if overwrite_existing:
            dst_path.unlink()
            shutil.copy2(src_path, dst_path)
            log(f"Overwrote copied raster: {dst_path}")
        else:
            log(f"Already exists, keeping: {dst_path}")
    else:
        shutil.copy2(src_path, dst_path)
        log(f"Copied: {dst_path}")

    written_rasters[var_name] = dst_path

if not written_rasters:
    raise RuntimeError("No rasters were found to copy/build into the NetCDF stack.")

# =============================================================================
# WRITE SIDEcar LOOKUP FILES
# =============================================================================

for var_name, class_map in CLASS_MAPS.items():
    if class_map:
        write_lookup_files(var_name, class_map, out_dir)

# =============================================================================
# EMBED CLASS MAPS INTO TIFF TAGS
# =============================================================================

for var_name, tif_path in written_rasters.items():
    if var_name in CLASS_MAPS and CLASS_MAPS[var_name]:
        embed_classmap_in_tif(tif_path, CLASS_MAPS[var_name])

# =============================================================================
# BUILD XARRAY DATASET
# =============================================================================

data_vars = {}

for var_name, tif_path in written_rasters.items():
    da = rxr.open_rasterio(tif_path, masked=True).squeeze("band", drop=True)
    da.name = var_name

    # Preserve source info
    da.attrs["source_tif"] = str(tif_path)

    # Attach class map where relevant
    if var_name in CLASS_MAPS and CLASS_MAPS[var_name]:
        da.attrs["class_map_json"] = json.dumps(CLASS_MAPS[var_name])

    data_vars[var_name] = da

ds = xr.Dataset(data_vars)

# Dataset-level metadata
example_da = next(iter(data_vars.values()))
ds.attrs["title"] = "LTDL-derived management covariate stack"
ds.attrs["source_directory"] = str(src_dir)
ds.attrs["written_directory"] = str(out_dir)
ds.attrs["crs"] = str(example_da.rio.crs)
ds.attrs["notes"] = (
    "Variables derived from LTDL treatment rasters aligned to the 10 m template grid. "
    "Categorical variables store label maps as JSON in variable attributes and sidecar CSV/JSON files."
)

# =============================================================================
# CAST VARIABLE DTYPES
# =============================================================================

for v in ds.data_vars:
    ds[v] = ds[v].astype(choose_dtype(v))

# =============================================================================
# WRITE SINGLE NETCDF WITH netCDF4 ENGINE
# =============================================================================

ensure_absent(nc_path)

encoding = {
    v: {
        "zlib": True,
        "complevel": 4,
        "dtype": choose_dtype(v)
    }
    for v in ds.data_vars
}

log("Writing NetCDF with engine='netcdf4' ...")
ds.to_netcdf(
    nc_path,
    format="NETCDF4",
    engine="netcdf4",
    encoding=encoding
)
log(f"Wrote NetCDF: {nc_path}")

# =============================================================================
# OPTIONAL: WRITE ONE NETCDF PER VARIABLE
# =============================================================================

if write_individual_netcdfs:
    for v in ds.data_vars:
        var_nc = out_dir / f"{v}.nc"
        ensure_absent(var_nc)

        ds[[v]].to_netcdf(
            var_nc,
            format="NETCDF4",
            engine="netcdf4",
            encoding={
                v: {
                    "zlib": True,
                    "complevel": 4,
                    "dtype": choose_dtype(v)
                }
            }
        )
        log(f"Wrote variable NetCDF: {var_nc}")

# =============================================================================
# VALIDATE OUTPUT
# =============================================================================

log("Re-opening NetCDF for validation ...")
ds_check = xr.open_dataset(nc_path, engine="netcdf4")

print("\nDataset summary:")
print(ds_check)

print("\nVariables:")
for v in ds_check.data_vars:
    print(f"- {v}: dtype={ds_check[v].dtype}, shape={ds_check[v].shape}")
    if "class_map_json" in ds_check[v].attrs:
        print("  has class_map_json")

print("\nDone.")
```
